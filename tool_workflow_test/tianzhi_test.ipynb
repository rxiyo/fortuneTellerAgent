## 八字计算工作流：基础命盘与按需计算

这一节展示不使用 Agent 的工具流：先排基础命盘，再根据问题决定是否计算原局规则和岁运。示例资料为虚构数据，不包含输入校验、自然语言解析、知识检索或最终解读。逐格顺序运行即可看到实际计算结果；后续问题可以复用同一个 `chart`。

示例约定出生时间为公历北京时间，传入北京经度并启用真太阳时；库的 `gender` 参数为 0 男、1 女，用于大运顺逆。不同排盘口径或输入会影响结果。旺衰、用神等是库按选定规则给出的结果，不能视为经过科学验证的预测。


### 1. 基础命盘：把出生时间转换成四柱

对应算命流程中的“排盘”。如果只问四柱，运行到这里就可以；更深入的问题再使用下方结果。


```python
from datetime import datetime
from pprint import pprint
from tianzhi_core import bazi

# 示例：公历北京时间 2000-01-01 12:00，北京经度约东经 116.4°。
# naive datetime 在本库中按东八区理解；经度由调用方提供，地点名称不会自动转经度。
birth_time = datetime(2004, 3, 7, 12, 0)
chart = bazi.chart.build_chart(
    birth_time,
    longitude=116.4,       # 用于真太阳时修正
    gender=1,              # 本库约定：0 男、1 女；影响大运顺逆
    use_true_solar=True,   # 使用真太阳时排盘
    late_zi="next_day",    # 晚子时换日口径
)

# 展示四柱、日主与计算口径；chart.quad 会作为后续规则计算的输入。
print("四柱（年月日时）：", chart.bazi)
print("日主及五行：", chart.day_master, chart.day_element)
print("北京时间：", chart.clock)
print("计算所用太阳时：", chart.solar)
print("太阳时修正分钟：", round(chart.solar_correction, 2))
print("晚子时口径：", chart.late_zi)
```

### 2. 按需查看十神与藏干

算命流程中，十神是其他天干相对于日主的关系；藏干是按规则从地支中取出的天干。它们为后续事业、感情、财富等主题提供线索。用户只问四柱时可以跳过这一步。


```python
# 遍历四柱：展示每柱的天干十神，以及地支藏干和藏干十神。
# 日柱天干本身作为参照点，在展示中标为“日主”。
for name, pillar in chart.named_pillars.items():
    gan_god = "日主" if name == "day" else chart.ten_god_of(name)
    print(f"{name}柱 {pillar.ganzhi}：天干十神={gan_god}；地支藏干=")
    pprint(chart.hidden(name), sort_dicts=False)
```

### 3. 按需计算原局规则：旺衰、格局和用神

例如用户要求事业或感情的总体分析，可以在基础命盘上进一步查看这些规则结果。以下按库的默认口径计算；输出是分析材料和规则依据，不能仅凭某一项直接推断现实事件。


```python
# 原局是出生时的四柱。先计算五行力量、日主旺衰及依据。
quad = chart.quad
print("五行力量（库的加权模型）：")
pprint(bazi.strength.element_power(quad), sort_dicts=False)
print("\n日主旺衰及依据：")
pprint(bazi.strength.day_master_strength(quad))

# 调候关注季节条件；月令格局与用神按库默认规则推导。
print("\n调候需求：")
pprint(bazi.tiaohou.climate_need(quad))
print("\n月令格局：")
pprint(bazi.geju.month_pattern(quad))
print("\n用神、喜忌及依据：")
pprint(bazi.yongshen.select(quad))
```

### 4. 按需计算大运和流年：示例问题“2027 年感情有哪些命理线索？”

这个问题带有时间，传统分析会把目标年份和原局、所处大运一起看。下方人为指定 2027 **立春流年**，并非自然语言解析；它从 2027 年立春开始，到 2028 年立春前结束。若问题指整个公历 2027 年，还需要覆盖立春前的上一流年。


```python
# 先排大运：库的性别和年干规则决定顺逆；sect=1 指定起运算法。
# 起运前不属于任何一步大运；每步结果的 start_year/end_year 只有年份精度。
start_age, first_transition = bazi.luck_cycle.start_age(chart, sect=1)
dayun = bazi.luck_cycle.dayun_list(chart, count=10, sect=1)
print("大运方向：", "顺行" if bazi.luck_cycle.is_forward(chart) else "逆行")
print("起运折算岁数：", start_age)
print("首次交运时间（库使用的排盘时间口径）：", first_transition)
print("各步大运（年龄为虚岁）：")
for cycle in dayun:
    pprint(cycle.to_dict(), sort_dicts=False)

# 查询指定立春流年的干支和虚岁。
target_year = 2027
liunian = bazi.luck_cycle.liunian(chart, target_year, target_year)[0]
print("\n目标流年：")
pprint(liunian.to_dict(), sort_dicts=False)

# 一个立春流年跨两个公历年。用年度范围保守地找可能相交的大运：
# 还包括结束于目标年前一年的运，避免遗漏目标年交运前仍有效的候选。
# 年度范围不提供每次交运的精确时刻，候选可能偏多，交界处需另外核实。
candidates = [c for c in dayun
              if c.start_year <= target_year + 1 and c.end_year >= target_year - 1]
print("\n可能相交的大运候选（年份粗筛，非精确交运判断）：")
for cycle in candidates:
    pprint(cycle.to_dict(), sort_dicts=False)

# 分别计算各候选大运、流年与原局的关系，保留库返回的全部依据。
# 未起运或无候选时，仍可单独计算流年与原局，不能假定已有大运。
if not candidates:
    print("\n无大运候选，先仅计算流年与原局：")
    pprint(bazi.interact.combine(quad, year_gz=liunian.ganzhi))
for cycle in candidates:
    print(f"\n流年 {liunian.ganzhi} + 大运候选 {cycle.ganzhi} 的关系：")
    pprint(bazi.interact.combine(quad, year_gz=liunian.ganzhi, dayun_gz=cycle.ganzhi))
```

### 5. 按需细看流月

用户进一步问某个时间段时，可以查看节气流月。这里只展示每个流月的干支和起止时间；它本身不是感情事件预测。流月按节气划分，不等同于公历月。


```python
# 每个区间包含开始时刻、不包含结束时刻；按需选择目标时间所在的流月。
print(f"{target_year} 立春流年的 12 个流月：")
for month in bazi.luck_cycle.liuyue(chart, target_year):
    print(f"{month.jieqi} {month.ganzhi}：{month.start} 至 {month.end}（结束不含）")

# 后续可结合用户问题、明确的命理规则与可追溯资料进行分析和表达。
# 原局、岁运与上述规则输出都不能直接证明现实事件何时发生。
```
